```python
#Importing PySpark Libraries

from pyspark.sql import functions as F
from pyspark.sql.window import Window
```

```python
# Bronze Layer – Reading the CSV from Storage

# Set the path of the uploaded file in Workspace / DBFS
file_path = "/Volumes/assesment_2/meta_customerchurm/meta_data/Customer_Churn_500k_split.csv"

# Read the provided CSV file
df_raw = (spark.read
          .option("header", "true")
          .option("inferSchema", "true")
          .option("sep", ",")
          .csv(file_path))

# Confirm data upload
print(f"Total number of records uploaded to the Bronze Tier: {df_raw.count():,}")
```

```python
#Silver Layer – Date and Time Standardisation and Rounding

# Reconstruct the native timestamp by concatenating Date and Time
df_prepared = df_raw.withColumn(
    "timestamp_clean", 
    F.to_timestamp(F.concat_ws(" ", F.col("Date"), F.col("Time")))
)

# Preview the chronological order by user
df_prepared.select("user_id", "Date", "Time", "timestamp_clean").show(5, truncate=False)
```

```python
# Configuring the distributed window and selecting metrics

# Define the user-specific distributed window specification
window_spec = Window.partitionBy("user_id").orderBy("timestamp_clean")

# Basic metrics on which we will generate the time series
metrics = ["avg_session_duration", "engagement_score", "revenue_per_user", "watch_time"]
```

```python

# Gold Layer – Vectorised Generation of Over 5,000 Features
features_exprs = []

for metric in metrics:
    # A. Time Lags: 1,000 per metric (Total: 4,000 columns)
    for offset in range(1, 1001):
        features_exprs.append(
            F.lag(F.col(metric), offset).over(window_spec).alias(f"{metric}_lag_{offset}")
        )
    
    # B. Moving Averages: 250 per metric (Total: 1,000 columns)
    for frame in range(2, 252):
        w_frame = window_spec.rowsBetween(-frame, 0)
        features_exprs.append(
            F.avg(F.col(metric)).over(w_frame).alias(f"{metric}_roll_avg_{frame}")
        )
    
    # C. Rolling standard deviations (Rolling StdDev): 100 per metric (Total: 400 columns)
    for frame in range(2, 102):
        w_frame = window_spec.rowsBetween(-frame, 0)
        features_exprs.append(
            F.stddev(F.col(metric)).over(w_frame).alias(f"{metric}_roll_std_{frame}")
        )

print(f"Total number of expressions created for the list: {len(features_exprs)}")
```

```python
#Unified Column Projection and Handling of Null Values

# Map all columns in a single transformation
df_5000_cols = df_prepared.select(*[F.col(c) for c in df_prepared.columns], *features_exprs)

# Fill in null values at the start of the window with 0
df_final = df_5000_cols.fillna(0)

# Check the final width of the DataFrame
print(f"✅ Final total number of columns in the table: {len(df_final.columns)}")
```

```python
# Name of the destination table in the Databricks catalogue
output_table_name = "customer_churn_5000_final"

# Writing to Delta Lake with partitioning by date
(df_final.write
 .format("delta")
 .mode("overwrite")
 .partitionBy("Date")
 .saveAsTable(output_table_name))

print(f"✅ Success: Table '{output_table_name}' stored correctly in Delta Lake.")
```

```python
#Part A: Design and Implementation of MapReduce Job for Temporal Analysis

# SECTION 1: DATA INGESTION & ENVIRONMENT SETUP

from pyspark.sql.functions import col, lit, sum as _sum, avg, round as _round, when

# Define Unity Catalog Volume path for Meta dataset
file_path = "/Volumes/assesment_2/meta_customerchurm/meta_data/Customer_Churn_500k_split.csv"

# Read CSV dataset from Unity Catalog Volume
df_meta = (spark.read
          .option("header", "true")
          .option("inferSchema", "true")
          .option("sep", ",")
          .csv(file_path))

# Verify schema and row count
print("=== METADATA DATASET SCHEMA ===")
df_meta.printSchema()
print(f"Total records ingested: {df_meta.count():,}")
```

```python

# SECTION 2: MAP PHASE (Spark-Native Implementation)
# Emitting (Key, Value) pairs: Key = Date, Value = (revenue, watch_time, count=1)

mapped_df = (
    df_meta
    .select(
        col("Date"),
        col("revenue_per_user").cast("double").alias("revenue"),
        col("watch_time").cast("double").alias("watch_time"),
        lit(1).alias("count")
    )
)

print("=== STAGE 1 (MAP): SCHEMA & PREVIEW ===")
mapped_df.printSchema()
mapped_df.show(5, truncate=False)
```

```python

# SECTION 3: REDUCE PHASE (Spark-Native Aggregation)
# Performing distributed reduction by Date key

reduced_df = (
    mapped_df
    .groupBy("Date")
    .agg(
        _sum("revenue").alias("total_revenue"),
        _sum("watch_time").alias("total_watch_time"),
        _sum("count").alias("total_interactions")
    )
)

print("=== STAGE 2 (REDUCE): SCHEMA & PREVIEW ===")
reduced_df.printSchema()
reduced_df.show(5, truncate=False)
```

```python


# SECTION 4: PROCESS & AVERAGE COMPUTATION
# Calculating daily average metrics (ARPU & Average Watch Time)

daily_meta_summary = (
    reduced_df
    .withColumn(
        "avg_revenue_per_user",
        when(col("total_interactions") > 0, col("total_revenue") / col("total_interactions")).otherwise(None)
    )
    .withColumn(
        "avg_watch_time",
        when(col("total_interactions") > 0, col("total_watch_time") / col("total_interactions")).otherwise(None)
    )
    .withColumn("avg_revenue_per_user", _round(col("avg_revenue_per_user"), 2))
    .withColumn("avg_watch_time", _round(col("avg_watch_time"), 2))
    .select("Date", "avg_revenue_per_user", "avg_watch_time", "total_interactions")
    .orderBy("Date")
)

print("=== STAGE 3 (FINAL OUTPUT): DAILY METRICS SUMMARY ===")
display(daily_meta_summary)
```

```python
# SECTION 5: ALTERNATIVE MAPREDUCE USING mapInPandas()
# Vectorized Map execution with Apache Arrow / Pandas UDF

def map_function(iterator):
    for pdf in iterator:
        pdf = pdf.copy()
        pdf['revenue'] = pdf['revenue_per_user'].astype(float)
        pdf['watch_time'] = pdf['watch_time'].astype(float)
        pdf['count'] = 1
        yield pdf[['Date', 'revenue', 'watch_time', 'count']]

# Map Phase via Arrow/Pandas
mapped_df_pandas = df_meta.mapInPandas(
    map_function,
    schema="Date date, revenue double, watch_time double, count int"
)

# Reduce Phase via Spark
reduced_df_pandas = (
    mapped_df_pandas
    .groupBy("Date")
    .agg(
        _sum("revenue").alias("total_revenue"),
        _sum("watch_time").alias("total_watch_time"),
        _sum("count").alias("total_interactions")
    )
)

# Post-processing Phase
daily_meta_pandas = (
    reduced_df_pandas
    .withColumn("avg_revenue_per_user", _round(col("total_revenue") / col("total_interactions"), 2))
    .withColumn("avg_watch_time", _round(col("total_watch_time") / col("total_interactions"), 2))
    .select("Date", "avg_revenue_per_user", "avg_watch_time", "total_interactions")
    .orderBy("Date")
)

print("=== MAPINPANDAS MAPREDUCE OUTPUT ===")
display(daily_meta_pandas)
```

```python
# SECTION 2: EXPLICIT REPARTITIONING BY DATE KEY

from pyspark.sql.functions import col

# Explicit repartitioning on the 'Date' key to balance worker memory during Reduce
mapped_repartitioned_df = mapped_df.repartition(col("Date"))

# Balanced reduce phase across cluster executors
reduced_balanced_df = (
    mapped_repartitioned_df
    .groupBy("Date")
    .agg(
        _sum("revenue").alias("total_revenue"),
        _sum("watch_time").alias("total_watch_time"),
        _sum("count").alias("total_interactions")
    )
)

print("=== REPARTITIONED REDUCE EXECUTION PLAN ===")
reduced_balanced_df.explain(mode="formatted")
```

```python
# Task 3: Implement Machine Learning Algorithms on Big Data for Predictions and Classification on Time Series Data
# Part A

# SECTION 1: DATA PREPARATION AND FEATURE VECTORIZATION

from pyspark.ml.feature import VectorAssembler
from pyspark.sql.functions import col

# Load the engineered dataset from Delta Lake Gold Layer
df_ml = spark.read.table("workspace.default.customer_churn_5000_final")

# Handle any remaining nulls or NaNs for ML compatibility
df_ml_clean = df_ml.na.fill(0)

# Define feature columns excluding identifiers, raw dates, and targets
exclude_cols = ["Date", "Time", "timestamp_clean", "user_id", "churn_flag"]
feature_cols = [c for c, dtype in df_ml_clean.dtypes if c not in exclude_cols and dtype in ('int', 'double', 'float', 'bigint', 'long', 'boolean')]

# Assemble input features into a single ML vector column
assembler = VectorAssembler(
    inputCols=feature_cols,
    outputCol="raw_features",
    handleInvalid="skip"
)

df_assembled = assembler.transform(df_ml_clean)

# Split dataset into Training (80%) and Testing (20%) sets for robust evaluation
train_data, test_data = df_assembled.randomSplit([0.8, 0.2], seed=42)

print(f"✅ Training Dataset Records: {train_data.count():,}")
print(f"✅ Testing Dataset Records: {test_data.count():,}")
```

```python

#  MODEL COMPARISON (Random Forest vs. GBT Classifier)

from pyspark.ml.classification import RandomForestClassifier, GBTClassifier
from pyspark.ml.evaluation import BinaryClassificationEvaluator
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

print("🚀 Starting model comparison: Random Forest vs. GBT Classifier...")

# 1. Train Random Forest Classifier (if not already trained)
rf = RandomForestClassifier(featuresCol="features", labelCol="churn_flag", numTrees=20, maxDepth=3)
rf_model = rf.fit(train_data)
rf_predictions = rf_model.transform(test_data)

# 2. Train Alternative Model: Gradient-Boosted Tree Classifier (GBTClassifier)
gbt_clf = GBTClassifier(featuresCol="features", labelCol="churn_flag", maxIter=10, maxDepth=3)
gbt_model = gbt_clf.fit(train_data)
gbt_predictions = gbt_model.transform(test_data)

# 3. Evaluate both models using ROC-AUC
evaluator = BinaryClassificationEvaluator(labelCol="churn_flag", rawPredictionCol="rawPrediction", metricName="areaUnderROC")

rf_auc = evaluator.evaluate(rf_predictions)
gbt_auc = evaluator.evaluate(gbt_predictions)

print(f"📊 Random Forest Test ROC-AUC: {rf_auc:.4f}")
print(f"📊 GBT Classifier Test ROC-AUC: {gbt_auc:.4f}")

# 4. Generate comparison DataFrame for plotting
comparison_data = pd.DataFrame({
    'Model': ['Random Forest', 'GBT Classifier'],
    'ROC-AUC Score': [rf_auc, gbt_auc]
})


# GENERATE VISUALIZATION CHARTS FOR THE PORTFOLIO

plt.figure(figsize=(8, 5))
sns.set_theme(style="whitegrid")
barplot = sns.barplot(x='Model', y='ROC-AUC Score', data=comparison_data, palette='Blues_d')

plt.title('Model Performance Comparison: ROC-AUC Score', fontsize=14, fontweight='bold')
plt.ylabel('ROC-AUC', fontsize=12)
plt.xlabel('Classification Algorithms', fontsize=12)
plt.ylim(0.4, 1.0)

# Add values on top of bars
for p in barplot.patches:
    barplot.annotate(format(p.get_height(), '.4f'), 
                   (p.get_x() + p.get_width() / 2., p.get_height()), 
                   ha = 'center', va = 'center', 
                   xytext = (0, 8), 
                   textcoords = 'offset points',
                   fontsize=11, fontweight='bold')

plt.tight_layout()
plt.show()

print("✅ Model comparison executed and evaluation chart generated successfully!")
```

```python
# 3. Classification and prediction implementation

from pyspark.ml.feature import VectorAssembler
from pyspark.ml.classification import RandomForestClassifier
from pyspark.ml.regression import GBTRegressor

# Load cleaned data
df_ml = spark.read.table("workspace.default.customer_churn_5000_final")

# Select only the original principal variables + a limited subset of lags 
base_cols = ["session_count", "avg_session_duration", "posts_created", "likes_given", 
             "comments_made", "shares", "watch_time", "revenue_per_user", "engagement_score"]

# Or dynamically filter columns that are not part of a mass expansion
selected_features = [c for c, dtype in df_ml.dtypes if c not in ["Date", "Time", "timestamp_clean", "user_id", "churn_flag"] and dtype in ('int', 'double', 'float', 'bigint', 'long', 'boolean')][:30]

print(f"Number of features selected for rapid training: {len(selected_features)}")

# Project only the columns needed for training and downstream visualization to reduce bytes read
needed_cols = selected_features + ["churn_flag", "timestamp_clean"]
df_ml = df_ml.select(*needed_cols).na.fill(0)

# Optimised vector assembly
assembler = VectorAssembler(inputCols=selected_features, outputCol="features", handleInvalid="skip")
df_assembled = assembler.transform(df_ml)

# Splitting data (training and testing)
train_data, test_data = df_assembled.randomSplit([0.8, 0.2], seed=42)

# Train an optimised Random Forest
rf = RandomForestClassifier(labelCol="churn_flag", featuresCol="features", numTrees=20, maxDepth=5, seed=42)
print("Training a Random Forest with bounded features...")
rf_model = rf.fit(train_data)
rf_predictions = rf_model.transform(test_data)
print("✅ Training successfully completed!")
```

```python

# SECTION 4: EVALUATION METRICS (OPTIMIZED & STABLE)

from pyspark.ml.evaluation import BinaryClassificationEvaluator, RegressionEvaluator
from pyspark.ml.regression import GBTRegressor

# --- Train GBT Regressor for watch_time prediction ---
gbt = GBTRegressor(labelCol="watch_time", featuresCol="features", maxDepth=3, maxIter=10, seed=42)
gbt_model = gbt.fit(train_data)
gbt_predictions = gbt_model.transform(test_data)

# --- Evaluation of Classification Model (Random Forest) ---
binary_evaluator = BinaryClassificationEvaluator(
    labelCol="churn_flag",
    rawPredictionCol="rawPrediction",
    metricName="areaUnderROC"
)

roc_auc = binary_evaluator.evaluate(rf_predictions)
print(f"📊 Random Forest Test ROC-AUC: {roc_auc:.4f}")

# --- Evaluation of Regression Model (GBT Regressor) ---
reg_evaluator = RegressionEvaluator(
    labelCol="watch_time",
    predictionCol="prediction",
    metricName="rmse"
)

rmse_val = reg_evaluator.evaluate(gbt_predictions)
print(f"📊 GBT Regressor Test RMSE: {rmse_val:.4f}")

print("✅ Model evaluation metrics successfully calculated!")
```
